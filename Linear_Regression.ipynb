```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score,mean_squared_error,mean_absolute_error
import matplotlib.pyplot as plt


df=pd.read_csv('/content/Housing.csv')
```

```python
df.head()
```

```python
df.tail()
```

```python
df=df.dropna()
```

```python
x=df.drop(['mainroad','guestroom','hotwaterheating','airconditioning','prefarea','furnishingstatus','basement','price'],axis=1)
y=df['price']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model=LinearRegression()
```

```python
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
r2=r2_score(y_test,y_pred)
m=mean_absolute_error(y_test,y_pred)
k=mean_squared_error(y_test,y_pred)
```

```python
plt.figure(figsize=(7,7))
plt.scatter(y_test,y_pred)
max_=y_test.max()
min_=y_test.min()
plt.plot([max_,min_],[max_,min_],color='red')
plt.xlabel("Actual")
plt.ylabel("Predicted")
plt.title("Linear Regression")
```

```python
print('R2 Score:',r2)
print('Mean Absolute Error: ',m)
print('Mean Squared Error: ',k)
```
