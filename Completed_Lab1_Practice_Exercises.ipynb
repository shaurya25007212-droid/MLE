**Q4.** Two arrays represent 5 students' marks in two tests: `test1 = np.array([65, 70, 55, 80, 90])` and

**Q9.** Create a new column `price_per_sqft` in `df` equal to `price_lakhs` (in lakhs) converted to rupees per
square foot. (1 lakh = 100,000 rupees, so `price_per_sqft = (price_lakhs * 100000) / area_sqft`.)
Display the 5 houses with the **highest** `price_per_sqft`.


````python
# Q9
df["price_per_sqft"] = (df["price_lakhs"] * 100000) / df["area_sqft"]

print(df.nlargest(5, "price_per_sqft")[["locality", "area_sqft", "price_lakhs", "price_per_sqft"]])
````

**Q10.** Using `.groupby()`, find the average `price_lakhs` and average `area_sqft` **per locality**, sorted from
highest to lowest average price.


````python
# Q10
locality_avg = (
    df.groupby("locality")[["price_lakhs", "area_sqft"]]
      .mean()
      .sort_values("price_lakhs", ascending=False)
)

print(locality_avg)
````

**Q11.** Count how many houses have **more than 2 bathrooms**, and separately, how many houses are **older than
15 years**.


````python
# Q11
print("Houses with more than 2 bathrooms:", (df["bathrooms"] > 2).sum())
print("Houses older than 15 years:", (df["age_years"] > 15).sum())
````

**Q12.** Create a new column `size_category` in `df`: `"Small"` if `area_sqft < 900`, `"Medium"` if
`900 <= area_sqft < 1400`, and `"Large"` otherwise. (Hint: `np.select` or a small function with `.apply()`.)


````python
# Q12
conditions = [
    df["area_sqft"] < 900,
    (df["area_sqft"] >= 900) & (df["area_sqft"] < 1400)
]
choices = ["Small", "Medium"]

df["size_category"] = np.select(conditions, choices, default="Large")
print(df[["area_sqft", "size_category"]])
````

**Q14.** Using `.corr()`, find which single numeric feature has the **strongest correlation with `price_lakhs`**.
Print the full correlation of every numeric column against `price_lakhs` only (not the whole matrix).


````python
# Q14
correlations = df.select_dtypes(include="number").corr()["price_lakhs"]
print(correlations)

strongest_feature = correlations.drop("price_lakhs").abs().idxmax()
print("Strongest correlation:", strongest_feature, "=", correlations[strongest_feature])
````

## Section C: Missing Values

**Q15.** Run the cell below to create `df_messy`, a version of the housing data with some missing values
inserted. Then print how many missing values exist in **each column**.


````python
# Run this first — do not modify
df_messy = df.copy()
df_messy.loc[[2, 7, 15], "area_sqft"] = np.nan
df_messy.loc[[1, 9], "bathrooms"] = np.nan
df_messy.loc[[4, 11, 18], "price_lakhs"] = np.nan
df_messy
````

````python
# Q15
print(df_messy.isna().sum())
````

**Q16.** Create `df_filled`, a copy of `df_messy` where:
- missing `area_sqft` values are filled with the **column median**
- missing `bathrooms` values are filled with the **column mode** (most frequent value)
- missing `price_lakhs` values are filled with the **mean price of the same locality** (hint: `.groupby()` +
  `.transform()`, or fill locality-by-locality)


````python
# Q16
df_filled = df_messy.copy()

df_filled["area_sqft"] = df_filled["area_sqft"].fillna(df_filled["area_sqft"].median())
df_filled["bathrooms"] = df_filled["bathrooms"].fillna(df_filled["bathrooms"].mode()[0])

locality_mean_price = df_filled.groupby("locality")["price_lakhs"].transform("mean")
df_filled["price_lakhs"] = df_filled["price_lakhs"].fillna(locality_mean_price)

print(df_filled)
````

**Q17.** Instead of filling, create `df_complete_only` by **dropping** any row in `df_messy` that has at least
one missing value. How many rows are left compared to the original 20?


````python
# Q17
df_complete_only = df_messy.dropna()

print("Rows left:", len(df_complete_only))
print("Rows removed:", len(df) - len(df_complete_only))
````

## Section D: Mini Challenge

**Q18.** Without using any missing-value dataset — using the original clean `df` — answer in one line of code
each:
- Which locality has the **most expensive average house**?
- Which locality has the **most number of houses listed**?
- What is the price range (max − min) of houses with **exactly 2 bedrooms**?


````python
# Q18
print(df.groupby("locality")["price_lakhs"].mean().idxmax())
print(df["locality"].value_counts().idxmax())
print(df.loc[df["bedrooms"] == 2, "price_lakhs"].max() - df.loc[df["bedrooms"] == 2, "price_lakhs"].min())
````

**Q19. (Challenge)** Write a single block of code that, for each locality, prints a one-line summary in this
exact format:

```
Malviya Nagar: 5 houses, avg price 55.4 lakhs, avg area 1040 sqft
```

(Hint: loop over `df["locality"].unique()`, or use `.groupby()` and iterate over the result with `.iterrows()`
on a summary DataFrame — both approaches are fine.)


````python
# Q19
summary = df.groupby("locality").agg(
    houses=("locality", "size"),
    avg_price=("price_lakhs", "mean"),
    avg_area=("area_sqft", "mean")
)

for locality, row in summary.iterrows():
    print(f"{locality}: {row['houses']} houses, avg price {row['avg_price']:.1f} lakhs, avg area {row['avg_area']:.0f} sqft")
````
